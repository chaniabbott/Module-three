```python
#read in my csv file - parsing the data

def read_books_no_csv(filename : str = 'books.csv'):
  #read books.csv withour using hte csv library,
  # returns a list of tuples where year is an int, so list will be
  # list[tuple{str,str,int,str,str}]

  fin = open(filename, 'r')
  lines = fin.readlines()
  fin.close()

  books = []
  for line in lines[1:]:
    line = line.strip()

    if line == '':
      continue

    #parts is a list
    parts = line.split(',')

    title = parts[0].strip()
    author = parts[1].strip()
    year = int(parts[2].strip())
    country = parts[3].strip()
    genre = parts[4].strip()

    books.append((title, author, year, country, genre ))

  return books


csv_data = read_books_no_csv()

for book in csv_data:
  print(f' one book{book}')
```

```python
def head(in_filename, n, out_filename=None):
 """
 Prints the first n lines of in_filename, or writes them to out_filename.
 If out_filename is None, prints to screen.
 """
 lines = []
 with open(in_filename, "r") as fin:
     for _ in range(n):
         line = fin.readline()
         if line == "":   # EOF
             break
         lines.append(line)


 if out_filename is None:
     for line in lines:
         print(line, end="")
 else:
     with open(out_filename, "w") as fout:
         for line in lines:
             fout.write(line)


head("books.csv", 5, None)
```

```python

def clean_to_words(line):
   # Replace common punctuation with spaces, then split
   punct = ".,;:!?\"'()[]{}-_—"
   line = line.lower()
   for thing in punct:
       line = line.replace(thing, " ")
   return line.split()


def is_pale_word(word):
   # matches pale, pales, paled, paleness
   if word.startswith("pale"):
       return True
   # matches pallor exactly
   if word == "pallor":
       return True
   return False


def count_pale_lines(filename):
   fin = open(filename, "r", encoding="utf-8")
   count = 0


   for line in fin:
       words = clean_to_words(line)
       found = False
       for w in words:
           if is_pale_word(w):
               found = True
               break
       if found:
           count += 1


   fin.close()
   return count
```
